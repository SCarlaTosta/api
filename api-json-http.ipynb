öncelikle bu çalişmamı veri tarafımı geliştirmek ve machine learning modelerimi iyileştirmek için yapıyorum umarım gerekli bilgimi ve denemelerimi burda etkili bi

evet pandasın data_frame fonku ile dash board üretiriz

```python
import pandas as pd
school = [
    {"öğrenci":"ahmet","grade":9},
    {"öğrenci":"mehmet","grade":10},
    {"öğrenci":"ayşe","grade":10},
    {"öğrenci":"kaan","grade":11},
    {"öğrenci":"elif","grade":10}
]

data_frame = pd.DataFrame(school)
print(data_frame)
```

.head ilk 5 basamaği yani satırı gösterir

```python
print(data_frame.head())
```

.shape ise şeklini

```python
print(data_frame.shape)
```

.colomns ise sütunları gösterir

```python
print(data_frame.columns)
```

.dtypes veri türlerini gösterir

```python
print(data_frame.dtypes)
```

.describe bize sayısaların analizini yapar

```python
print(data_frame.describe())
```

indlude= ""   bize hangi veri türlerinin analiz edilceni gösterir

```python
print(data_frame.describe(include="str"))
```

x[""] bize sütun seçmemizi sağlar


```python
print(data_frame["grade"])
```

[["",""]] yapısı bize daha fazla sütun seçmemizi sağlar

```python
print(data_frame[["öğrenci","grade"]])
```

x[""] == x    boylece karşılaştırma yapıp indexleri True - False ile çıkartabiliyoruz

```python
print(data_frame["grade"] == 10)
```

x[koşul][""] ile istediğiniz sütundan sadece belirli kurala göre seçebilirsiniz

```python
print(data_frame[data_frame["grade"] == 10]["öğrenci"])
```

.loc da aynı şekilde koşulu bi şekilde seçebilir [koşul,sütun]şeklinde

```python
print(data_frame.loc[data_frame["grade"] == 10,["öğrenci","grade"]])
```

x["new"] şeklinde yeni sütunlar oluşturabilirsiniz ve koşular sayesinde özeleştirebilirsiniz

```python
data_frame["durum"] = data_frame["grade"] >= 10
print(data_frame)
```

mevcut sütunuzu da değiştirebilir yada onu manipule edebilirsiniz

```python
data_frame["grade"] = data_frame["grade"] +1
print(data_frame)
```
