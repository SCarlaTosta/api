```python
import pandas as pd
school = [
    {"öğrenci":"ahmet","grade":9},
    {"öğrenci":"mehmet","grade":10},
    {"öğrenci":"ayşe","grade":10},
    {"öğrenci":"kaan","grade":11},
    {"öğrenci":"elif","grade":10}
]

data_frame = pd.DataFrame(school)
print(data_frame)
```

```python
print(data_frame[["öğrenci","grade"]])
```

```python
print(data_frame["grade"] == 10)
```

```python
print(data_frame[data_frame["grade"] == 10]["öğrenci"])
```

```python
print(data_frame.loc[data_frame["grade"] == 10,["öğrenci","grade"]])
```

```python
data_frame["durum"] = data_frame["grade"] >= 10
print(data_frame)
```
